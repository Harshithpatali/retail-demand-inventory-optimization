Explore how service-level and lead-time assumptions change safety stock and target inventory position.

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, ipywidgets as widgets
from src.inventory.safety_stock import safety_stock
mean=10; sd=2; review=7
service_levels=np.array([.80,.85,.90,.92,.95,.97]); rows=[{"service_level":s,"safety_stock":safety_stock(sd,7+review,s)} for s in service_levels]; display(pd.DataFrame(rows))
```

```python
fig,ax=plt.subplots(figsize=(8,4)); pd.DataFrame(rows).plot(x="service_level",y="safety_stock",marker="o",ax=ax,legend=False); ax.set_title("Safety stock vs service level"); plt.show()
```

```python
lead_times=[1,3,5,7,10,14,21,30]; rows2=[]
for L in lead_times:
    prot=L+review; ss=safety_stock(sd,prot,.90); rows2.append({"lead_time":L,"protection_period":prot,"safety_stock":ss,"S":mean*prot+ss})
display(pd.DataFrame(rows2))
```

```python
def sensitivity(sl,review_days):
    ss=safety_stock(sd,7+int(review_days),float(sl)); print({"service_level":sl,"review_period":review_days,"protection_period":7+int(review_days),"safety_stock":ss,"target_S":mean*(7+int(review_days))+ss})
widgets.interact(sensitivity,sl=widgets.FloatSlider(value=.90,min=.80,max=.97,step=.01),review_days=widgets.IntSlider(value=7,min=1,max=14,step=1))
```
