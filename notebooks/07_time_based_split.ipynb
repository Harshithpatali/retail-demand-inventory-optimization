# 07 — Time-Based Split

Random splits are inappropriate for forecasting. This notebook makes the chronological boundaries visible and verifies there is no overlap.

```python

from pathlib import Path
import os, sys
ROOT = Path.cwd()
while not (ROOT / "README.md").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
os.chdir(ROOT)
print("Project root:", ROOT)
```

```python
import pandas as pd
from src.utils.config import load_yaml
df=pd.read_parquet(ROOT/"data/processed/forecast_features.parquet"); cfg=load_yaml("model_config.yaml"); maxd=df.date.max(); test_start=maxd-pd.Timedelta(days=cfg["test_days"]-1); val_start=test_start-pd.Timedelta(days=cfg["validation_days"]); train=df[df.date<val_start]; val=df[(df.date>=val_start)&(df.date<test_start)]; test=df[df.date>=test_start]
print({"train":(train.date.min(),train.date.max(),len(train)),"validation":(val.date.min(),val.date.max(),len(val)),"test":(test.date.min(),test.date.max(),len(test))})
```

```python
print("Train/validation overlap:",len(set(train.date)&set(val.date))); print("Validation/test overlap:",len(set(val.date)&set(test.date))); print("Series retained in test:",test.id.nunique())
```

The final test window remains untouched until final comparison. This structure supports honest out-of-sample evaluation.
