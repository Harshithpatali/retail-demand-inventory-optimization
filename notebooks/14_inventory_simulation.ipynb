# 14 — Inventory Simulation

Trace the daily physical inventory process: arrivals, demand fulfillment, stockouts, review decisions and costs.

```python

from pathlib import Path
import os, sys
ROOT = Path.cwd()
while not (ROOT / "README.md").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
os.chdir(ROOT)
print("Project root:", ROOT)
```

```python
import pandas as pd
sim=pd.read_parquet(ROOT/"data/processed/inventory_simulation.parquet"); print(sim.shape); display(sim.head())
```

```python
for policy in ["baseline","optimized"]:
    q=sim[sim.policy==policy]; print("\n",policy); display(q[["date","demand","fulfilled","ending_inventory","inventory_position","order_quantity_placed","is_review_day","total_cost"]].head(20))
```

```python
import matplotlib.pyplot as plt
q=sim[sim.policy=="optimized"].groupby("date",observed=True).ending_inventory.mean(); fig,ax=plt.subplots(figsize=(10,4)); q.plot(ax=ax); ax.set_title("Average ending inventory — optimized policy"); plt.show()
```

### Fairness check

Both policies use the identical held-out dates and selected series. Review decisions are evaluated only on scheduled review days. Purchase orders arrive only after lead time.
