# 04 — Exploratory Data Analysis

Focus: distribution shape, zeros, time trend, product/store heterogeneity, and price behavior.

```python

from pathlib import Path
import os, sys
ROOT = Path.cwd()
while not (ROOT / "README.md").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
os.chdir(ROOT)
print("Project root:", ROOT)
```

```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
path=ROOT/"data/interim/cleaned_sales.parquet"
df=pd.read_parquet(path)
summary={"total_units":int(df.demand.sum()),"mean":float(df.demand.mean()),"median":float(df.demand.median()),"zero_rate":float((df.demand==0).mean()),"max":int(df.demand.max()),"series":int(df.id.nunique())}
print(summary)
```

```python
daily=df.groupby("date",observed=True).demand.sum(); fig,ax=plt.subplots(figsize=(10,4)); daily.plot(ax=ax); ax.set_title("Total daily demand"); ax.set_ylabel("Units"); plt.show()
```

```python
fig,ax=plt.subplots(figsize=(8,4)); ax.hist(df.demand.clip(upper=df.demand.quantile(.995)),bins=40); ax.set_title("Demand distribution (99.5th percentile capped for visibility)"); ax.set_xlabel("Units"); plt.show()
```

```python
series=df.groupby("id",observed=True).demand.agg(total="sum",zero_rate=lambda s:(s==0).mean(),mean="mean",max="max"); fig,ax=plt.subplots(figsize=(8,4)); ax.hist(series.zero_rate,bins=30); ax.set_title("Series zero-demand rate"); ax.set_xlabel("Zero-demand fraction"); plt.show(); display(series.sort_values("total",ascending=False).head(10))
```

```python
price=df.groupby("date",observed=True).sell_price.mean(); fig,ax=plt.subplots(figsize=(10,4)); price.plot(ax=ax); ax.set_title("Average observed price"); plt.show()
```
