Bring all five forecasting approaches into one comparable table and analyze performance by demand segment.

```python

from pathlib import Path
import os, sys
ROOT = Path.cwd()
while not (ROOT / "README.md").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
os.chdir(ROOT)
print("Project root:", ROOT)
```

```python
import pandas as pd
comp=pd.read_csv(ROOT/"reports/tables/model_comparison_validation.csv"); display(comp.sort_values("WAPE"))
```

```python
import matplotlib.pyplot as plt
fig,ax=plt.subplots(figsize=(8,4)); comp.sort_values("WAPE").plot(x="model",y="WAPE",kind="bar",ax=ax,legend=False); ax.set_title("Validation WAPE by model"); ax.set_ylabel("WAPE"); plt.xticks(rotation=30); plt.show()
```

```python
best=comp[comp.model.isin(["xgboost","lightgbm"])].sort_values("WAPE").iloc[0]; print("Selected ML model:",best.model); print("Primary metric:",best.WAPE)
```

### Decision rule

The ML winner is selected on validation WAPE. The final model is then refit on train+validation before being used for future forecasting. This keeps selection separate from the final test evaluation.
