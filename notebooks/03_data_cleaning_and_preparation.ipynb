# 03 — Data Cleaning and Preparation

The key design constraint is memory. We avoid a full-panel melt and work with a deterministic subset.

```python

from pathlib import Path
import os, sys
ROOT = Path.cwd()
while not (ROOT / "README.md").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
os.chdir(ROOT)
print("Project root:", ROOT)
```

```python
import pandas as pd
from src.utils.config import load_yaml
path=ROOT/"data/interim/cleaned_sales.parquet"
if path.exists():
    df=pd.read_parquet(path); print("Rows:",len(df)); print("Series:",df.id.nunique()); print("Memory MB:",df.memory_usage(deep=True).sum()/1024**2); display(df.head())
else: print("Run prepare_data first.")
```

```python
if path.exists():
    qa=(df.groupby("id",observed=True).demand.agg(total_demand="sum",mean_demand="mean",max_demand="max",zero_rate=lambda s:(s==0).mean()).reset_index())
    print("Completely zero series:",(qa.total_demand==0).sum()); display(qa.describe().T)
```

```python
if path.exists():
    store_counts=df.groupby("store_id",observed=True).id.nunique().sort_values(ascending=False); display(store_counts.to_frame("series_count"))
```

### Interpretation

The cleaned table is the reusable analytical backbone. The portfolio restriction is explicit rather than hidden: results describe the selected high-demand series, not the entire M5 panel.
