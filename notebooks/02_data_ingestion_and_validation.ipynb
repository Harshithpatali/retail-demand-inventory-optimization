# 02 — Data Ingestion and Validation

This notebook documents file-level and table-level validation before modeling.

```python

from pathlib import Path
import os, sys
ROOT = Path.cwd()
while not (ROOT / "README.md").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
os.chdir(ROOT)
print("Project root:", ROOT)
```

```python
import pandas as pd, json
from pathlib import Path
raw=ROOT/"data/raw"
files={p.name:p.exists() for p in raw.glob("*.csv")}
print(pd.Series(files).to_string())
```

```python
calendar_path=raw/"calendar.csv"; prices_path=raw/"sell_prices.csv"
if calendar_path.exists():
    calendar=pd.read_csv(calendar_path); calendar["date"]=pd.to_datetime(calendar["date"]); print(calendar.shape); display(calendar.isna().mean().sort_values(ascending=False).head(15).to_frame("missing_rate"))
if prices_path.exists():
    prices=pd.read_csv(prices_path); print(prices.shape); display(prices.describe(include="all").T.head(15))
```

```python
report_path=ROOT/"data/interim/validation_report.json"
if report_path.exists(): print(json.loads(report_path.read_text()))
else: print("Run prepare_data first to create the validation report.")
```
