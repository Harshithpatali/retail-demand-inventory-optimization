# 15 — Business Impact Analysis

Separate model quality from business outcomes. This notebook reports simulated cost changes without fabricating real-world savings.

```python

from pathlib import Path
import os, sys
ROOT = Path.cwd()
while not (ROOT / "README.md").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
os.chdir(ROOT)
print("Project root:", ROOT)
```

```python
import pandas as pd
impact=pd.read_csv(ROOT/"reports/tables/business_impact_summary.csv"); display(impact)
```

```python
import matplotlib.pyplot as plt
row=impact[impact.metric=="total_cost"].iloc[0]; print(f"Total-cost change: {row.change_pct:.2f}%")
fig,ax=plt.subplots(figsize=(7,4)); ax.bar(["Baseline","Optimized"],[row.baseline,row.optimized]); ax.set_title("Total simulated cost"); plt.show()
```

```python
for m in ["holding_cost","ordering_cost","stockout_cost","service_level","fill_rate"]:
    r=impact[impact.metric==m].iloc[0]; print(f"{m}: baseline={r.baseline:.3f}, optimized={r.optimized:.3f}, change={r.change_pct:.2f}%")
```

Interpretation should ask **why** total cost moved: holding, ordering and stockout costs can move in different directions. Service level is not optimized automatically unless the cost structure supports it.
